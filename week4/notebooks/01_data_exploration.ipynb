## Step 1: Load Raw Data from Unity Catalog Volume

In this step, we are loading the raw dataset from a Unity Catalog volume into a Spark DataFrame.

The dataset was previously uploaded to a managed volume, which represents the raw data layer in a Lakehouse architecture. By reading the file into a DataFrame, we enable distributed data processing using Apache Spark.

We use the following options:
- `header=True` to interpret the first row as column names
- `inferSchema=True` to automatically detect appropriate data types

This step represents the data ingestion phase of a data pipeline, where raw data is brought into the processing environment.

```python
df = spark.read.csv(
    "/Volumes/week4_catalog/default/week4_volume/sales_data.csv",
    header=True,
    inferSchema=True
)
```
